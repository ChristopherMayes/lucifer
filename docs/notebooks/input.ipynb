# The input file

`lucifer.Input` is the input file as Python, generated from the schema the program writes.

```python
from IPython.display import Code

from lucifer import Input, schema
```

## Build one

```python
inp = Input(lattice="aramis.bmad")
inp.beam.n_particle = 4096
inp["run.out_root"] = "s7"
inp
```

The TOML the program reads holds the lattice and the keys set away from their defaults.

```python
Code(inp.toml(), language="toml")
```

```python
inp.items()
```

## The schema

Each key's type, default, bounds and meaning, as the program states them.

```python
schema.keys()["beam"]["n_particle"]
```

```python
schema.keys()["run"]["source_filter_tolerance"]
```

## Flat text

`lucifer.flat` routes `key = value` lines into their tables, the form the check scripts write.

```python
from lucifer.flat import to_toml

Code(to_toml("lattice = 'aramis.bmad'\nn_particle = 2048\nlambda0 = 1e-10\n"), language="toml")
```
